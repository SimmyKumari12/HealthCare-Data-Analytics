Data cleaning

1. Cleaning the duplicate values

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

2.Cleaning Name Columns

```python
df['Name'] = df['Name'].str.strip().str.title()
```

```python
df['Name']
```

3. Fix Negative billing amounts

```python
df.loc[df['Billing Amount']<0, 'Billing Amount'] = df['Billing Amount'].mean()
```

```python
df.describe()
```

```python
df.head()
```

Create a new column of days of stay

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
```

```python
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
```

```python
df["Days of Stay"] = (df['Discharge Date'] - df['Date of Admission']).dt.days
```

Data Analysis and visualization

Questions to answer

1.Which medical condition affects the most patients across the network?

2. How is patient age distribued acrodd the network?

3. How has average treatment cost changed over time

```python
condition_count = df['Medical Condition'].value_counts()
condition_count
```

```python
plt.figure(figsize=(8,5))
plt.bar(condition_count.index, condition_count.values,color="red")
plt.xlabel('Medical Condition')
plt.ylabel('Number of patients')
plt.xticks(rotation = 90)
plt.title('Medical Condition Count')
```

How is patient age distributed across the network?

```python
avg_age = df["Age"].mean()
avg_age
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Age'],bins=20,color='red',edgecolor='black')
plt.title('Age distribution of Patients')
plt.ylabel("No of patients")
plt.xlabel('Age')
plt.tight_layout()
plt.show()
```

How has average treatment cost changes over time

```python
avg_cost = df['Billing Amount'].mean()
avg_cost
```

```python
df['Admission_month'] = df['Date of Admission'].dt.to_period('M')
monthly_avg = df.groupby('Admission_month')['Billing Amount'].mean()
monthly_avg
```

```python
plt.figure(figsize = (10,5))
plt.plot(monthly_avg.index.astype(str),monthly_avg.values,color = 'red',marker = 'o',markersize = 3)
plt.xlabel("Month")
plt.ylabel("Average Billing Amount")
plt.xticks(rotation = 90)
plt.title("Average Treatment Cost Over Time(Monthly)")
plt.show()
```

Relationship between patient age and treatment cost

```python
relAgeCost = df.groupby('Age')["Billing Amount"].mean()
relAgeCost
```

```python
plt.figure(figsize=(10,5))
plt.scatter(relAgeCost.index,relAgeCost.values,color="red",alpha = 0.5,s = 15)
plt.xlabel("Age")
plt.ylabel("Average Cost")
plt.xticks(rotation = 90)
plt.title("Relationship between Age and Treatment Cost")
plt.show()
```

Which insurance provider cover the most patients?

```python
insurance_count = df["Insurance Provider"].value_counts()
insurance_count
```

```python
plt.figure(figsize=(7,7))
plt.pie(insurance_count.values,labels=insurance_count.index,autopct = '%1.1f%%',colors=['Red','Green','Purple','Blue','Yellow'])
plt.title("Share of insurance proider w.r.t patients")
plt.show()
```
